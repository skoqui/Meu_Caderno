# Anotação: Aula 8 (Hello World)

```c
#include <stdio.h>
```

- Diz ao compilador para incluir a **biblioteca padrão de entrada e saída**.
- É ela que permite usar funções como:

    - `printf()` → imprimir texto na tela
    - `scanf()` → ler dados do usuário

→ Sem isso, o `printf` não funciona.

```c
int main() {
```

- `main` é a **função principal** do programa.
- Todo programa em C **começa a executar** por ela.
- `int` significa que essa função vai **retornar um número inteiro** no final.

As `{ }` indicam `onde o programa começa e termina`.

```c
printf("Hello World!\n");
```

- `printf` serve para **mostrar algo na tela**.
- `"Hello World!"` é o texto que será exibido.
- `\n` significa pular linha (quebra de linha).
- `O ;` indica o fim do comando.

Quando o programa roda, ele escreve:

````python
Hello World!
````

```c
return 0;
```

- Indica que o programa **terminou com sucesso**.
- `0` é o código padrão para dizer ao sistema:
    “Deu tudo certo.”

Boa prática sempre usar isso.

}

- Fecha a função `main.`
- Aqui o programa termina.

### Resumo rápido

- `#include` → importa biblioteca
- `main()` → ponto inicial do programa
- `printf()` → imprime texto
- `return 0` → finaliza corretamente
