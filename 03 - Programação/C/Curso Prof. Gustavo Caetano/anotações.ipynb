# Lógica de Programação e Algoritmos com C


### Lógica estrutural

**Exemplo 1:**
```
Início
    Escreva("Está chovendo?")
    Leia(resposta)
    Se resposta = "sim" então
        Escreva("Leve o guarda-chuva")
    Senão
        Escreva("Pode sair de boa pq não vai chover")
Fim
```

**Exemplo 2:**

```
Inicio
    Escreva("Você tem miojo? (Sim/Não)")
    Leia(temMiojo)
    Se temMiojo = "sim" então
        Escreva("Coloque água na panela")
        ferveu <- falso
        Enquanto ferveu = falso faça
            Escreva("Esperando ferver...")
            Escreva("A água ferveu? (sim/não)")
            Leia(resposta)
            Se resposta = "sim" então
                ferveu <- verdadeiro
        Escreva("Coloque o miojo e cozinhe por 3 minutos")
    Senão
        Escreva("Vai de pão com ovo hoje mesmo...")
Fim

```
