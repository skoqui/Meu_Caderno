### Função + contagem regressiva

```python
from time import sleep

def contagem(num):
    print(f'A contagem de {num} segundos será iniciada.')
    print('=='*20)
    sleep(2)
    for i in range(num, 0, -1):
        print(i)
        sleep(1)

    print(f'A contagem de {num} segundos chegou ao fim. 🎉🎉🎉')


contagem(10)
```

### Split() + texto invertido

```python
texto = "Python é muito interessante"
palavras = texto.split()
print(palavras)
```

```python
texto = "Python é muito interessante"
palavras = texto.split()
textoInvertido = " ".join(palavras[::-1])
print(textoInvertido)
```

# Loop While


```python
moviesList = ["Batman", "Vingadores", "Matrix"]

index = 0
while index < len(moviesList):
    print(moviesList[index])
    index += 1
```

### Explicação rápida:

- `index = 0` → começa do primeiro item da lista
- `len(moviesList)` → mostra quantos itens existem
- `while index < len(moviesList):` → repete enquanto index for menor que o tamanho da lista
- `print(moviesList[index])` → mostra o item atual
- `index += 1` → passa pro próximo item
